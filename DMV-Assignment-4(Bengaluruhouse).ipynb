```python
# 1. Import libraries
import pandas as pd
```

```python
# 2. Load dataset
df = pd.read_csv("Bengaluru_House_Data.csv")
```

```python
# Clean column names
df.columns = df.columns.str.strip().str.replace(' ', '_')
```

```python
# 3. Explore data
print(df.head())
df.info()
print(df.isnull().sum())
```

```python
# 4. Handle missing values
df = df.dropna(subset=['location'])
df['society'] = df['society'].fillna('Unknown')
df['size'] = df['size'].fillna('Unknown')
df['bath'] = df['bath'].fillna(df['bath'].median())
df['balcony'] = df['balcony'].fillna(df['balcony'].median())
```

```python
# 5. Convert total_sqft into numeric values
# This code converts the total_sqft column into numeric values. If a value is a range, it calculates its average. Invalid values are replaced with None.
def convert(x):
    try:
        return float(x)
    except:
        if '-' in str(x):
            a, b = x.split('-')
            return (float(a) + float(b)) / 2
        return None

df['total_sqft'] = df['total_sqft'].apply(convert)
```

```python
# Fill missing square feet values
df['total_sqft'] = df['total_sqft'].fillna(df['total_sqft'].median())
```

```python
# 6. Merge additional dataset (if available)
# extra = pd.read_csv("Neighborhood_Data.csv")
# df = pd.merge(df, extra, on='location', how='left')
```

```python
# 7. Filter data
df = df[df['price'] > 0]
df = df[df['total_sqft'] > 0]
```

```python
# 8. Encode categorical columns
#df = pd.get_dummies(df, columns=['area_type'], dtype=int)
print(df.columns.tolist())
```

```python
# 9. Aggregate data
print(df.groupby('location')['price'].mean())
```

```python
# 10. Handle outliers
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

df['price'] = df['price'].clip(Q1-1.5*IQR, Q3+1.5*IQR)
```

```python
# 11. Display final data
print(df.head())
print("Final Shape:", df.shape)
```

```python
# 12. Export cleaned data
df.to_csv("Cleaned_RealEstate_Data.csv", index=False)
print("Data saved successfully!")
```
