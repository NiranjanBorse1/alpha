```python
# 1. Import libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 2. Load dataset
df = pd.read_csv("Supermart Grocery Sales - Retail Analytics Dataset.csv")
```

```python
# 3. Explore dataset
print(df.head())
df.info()
print(df.isnull().sum())
```

```python
# 4. Group sales by region
region_sales = df.groupby('Region')['Sales'].sum()
print("Total Sales by Region:")
print(region_sales)
```

```python
# 5. Visualize sales by region
region_sales.plot(kind='bar')
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()

# Pie chart
region_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales Distribution by Region")
plt.ylabel("")
plt.show()
```

```python
# 6. Find top-performing region
#idxmax=Finds the region with highest sales
print("Top Performing Region:", region_sales.idxmax())
print("Highest Sales:", region_sales.max())
```

```python
# 7. Group sales by region and category
category_sales = df.groupby(
    ['Region', 'Category']
)['Sales'].sum()

print("Sales by Region and Category:")
print(category_sales)
```

```python
# 8. Create stacked bar plot
#unstack()-Converts categories into separate columns
category_sales.unstack().plot(kind='bar', stacked=True)
plt.title("Sales by Region and Category")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.legend(title="Category")
plt.show()
```
