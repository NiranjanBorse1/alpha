```python
# 1. Import libraries
import numpy as np
import random

# 2. Check winner
def winner(b, p):
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,4,6)]
    return any(all(b[i] == p for i in w) for w in wins)

# 3. Initialize Q-table
Q = {}
alpha = 0.1
gamma = 0.9
epsilon = 0.1

# 4. Train computer
for episode in range(5000):
    b = [' ']*9

    while True:
        state = ''.join(b)
        actions = [i for i in range(9) if b[i] == ' ']

        if state not in Q:
            Q[state] = np.zeros(9)

        if random.random() < epsilon:
            action = random.choice(actions)
        else:
            action = max(actions, key=lambda a: Q[state][a])

        b[action] = 'X'

        if winner(b, 'X'):
            reward = 1
            done = True
        elif ' ' not in b:
            reward = 0
            done = True
        else:
            opp = random.choice([i for i in range(9) if b[i] == ' '])
            b[opp] = 'O'

            done = winner(b, 'O')
            reward = -1 if done else 0

        next_state = ''.join(b)

        if next_state not in Q:
            Q[next_state] = np.zeros(9)

        next_actions = [i for i in range(9) if b[i] == ' ']

        future = max([Q[next_state][i] for i in next_actions], default=0)

        Q[state][action] += alpha * (
            reward + gamma * future - Q[state][action]
        )

        if done or ' ' not in b:
            break

# 5. Display board
def display(b):
    print()
    for i in range(0, 9, 3):
        print(b[i], '|', b[i+1], '|', b[i+2])
    print()

# 6. Play against computer
b = [' ']*9

while True:
    state = ''.join(b)
    actions = [i for i in range(9) if b[i] == ' ']

    if state not in Q:
        action = random.choice(actions)
    else:
        action = max(actions, key=lambda a: Q[state][a])

    b[action] = 'X'
    print("Computer played:")
    display(b)

    if winner(b, 'X'):
        print("Computer Wins!")
        break

    if ' ' not in b:
        print("Draw!")
        break

    # User input
    pos = int(input("Enter position (1-9): ")) - 1

    if pos not in range(9) or b[pos] != ' ':
        print("Invalid position! Try again.")
        b[action] = ' '
        continue

    b[pos] = 'O'
    display(b)

    if winner(b, 'O'):
        print("You Win!")
        break

    if ' ' not in b:
        print("Draw!")
        break
```
